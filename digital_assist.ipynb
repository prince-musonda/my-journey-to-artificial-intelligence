```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
model_name = "GEB-AGI/geb-1.3b"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

input_ = tokenizer("hello there, my name is Prince. and your name is Rio. will act as my assist because you are variation of iron man's Javis. ", return_tensors="pt")
print(input_)
```

```python
response = model.generate(**input_)
print(tokenizer.decode(response[0]))
```
