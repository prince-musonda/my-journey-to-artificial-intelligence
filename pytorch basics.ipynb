```python
heights = [189, 170, 189, 163, 183, 171, 185,
 168, 173, 183, 173, 173, 175, 178,
 183, 193, 178, 173, 174, 183, 183,
 180, 168, 180, 170, 178, 182, 180,
 183, 178, 182, 188, 175, 179, 183,
 193, 182, 183, 177, 185, 188, 188,
 182, 185, 191, 183]
```

```python
import torch

heights_tensor = torch.tensor(heights, dtype=torch.float64)
```

```python
tensor1 = torch.zeros(2,3)
print(tensor1)
```

```python
tensor2 = torch.ones(3,4,5)
print(tensor2)
```

```python
first_five = heights_tensor[:5]
print(first_five)
```

```python
feets_in_height = heights_tensor / 30.48
feets_in_height
```

```python
heights_2_measures = torch.cat([heights_tensor, feets_in_height],dim=-1)
heights_2_measures
```

```python
heights_reshaped = heights_2_measures.reshape((2,-1))
print(heights_reshaped)
```

```python
uncle_joe_height = heights_reshaped[0,-1]
```

```python
print(f"uncle Joe's height in centimeters is {uncle_joe_height}")
```

```python
median_height = torch.median(heights_reshaped[0])
print(f"the median height is {median_height}")
```

```python
average_height = torch.mean(heights_reshaped,dim=1)
print(average_height)
```

```python
# tallest president
values, indices = torch.max(heights_reshaped,dim=1)
print(values)
print(indices)
```
